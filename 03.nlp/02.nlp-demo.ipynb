# 1. Preprocessing

## 1.1. Tokenization

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer, LancasterStemmer
import string

# Download necessary nltk resources
nltk.download('punkt') #punctuator kit
nltk.download('punkt_tab')
nltk.download('stopwords') #pre-defined set of words that should not be embedded.
nltk.download('wordnet') #lexical database of english

text_input = """
Artificial Intelligence, often called AI, is the science of creating machines that can perform tasks requiring human-like intelligence.
These tasks include learning from data, recognizing speech, understanding natural language, and making decisions.
Machine Learning, a subset of AI, allows systems to improve automatically through experience without being explicitly programmed.
Deep Learning, which uses neural networks, has made breakthroughs in image recognition, self-driving cars, and chatbots.
Although AI brings enormous benefits, such as faster problem solving and medical advancements, it also raises ethical concerns like job automation and data privacy.
The future of AI may lead to general intelligence, where machines can perform any task a human can, but this is still an open research challenge.
"""

# Sentense Tokenizer
documents = nltk.sent_tokenize(text_input)
for document in documents:
    print(document)
    print()

# Word tokenizer
words = nltk.word_tokenize(text_input)
print(words)
print()

# Remove stopwords (remove common words like and, is, space, the, etc...)
stop_words = set(stopwords.words('english'))
filtered_words = [word.lower() for word in words if word.lower() not in stop_words]
print(filtered_words)
print(len(filtered_words))
print()
```

## 1.2. Text Normalization: Removing Special Characters

```python
import string
special_characters = set(string.punctuation)
print(special_characters)

print(filtered_words)
print()

# Remove special characters
filtered_words_without_special_characters = [word for word in filtered_words if word not in special_characters]
print(filtered_words_without_special_characters)
print()
```

## 1.3. Text Normalization: Stemming
Stemming (reduce the words to common base known as stems). E.g. running --> run, ran --> run, flies --> fli, happily --> happi

### Rule-based Stemming
Removing prefixes and suffixes from the words to find their root(stem).
Running > remove ing > run
Happily > remove ly > happi
Unhappy > remove un > happy

### Porter Stemming (most widely used)
- Like rule-based, it removes suffix, and applies grammer rule.

## Lancaster Stemming
- Aggressive form of Porter Stemming.
- It reduces the word too much that the word could sometime lose meaning.
- But it has better grammer rules compared Porter Stemming.

### Snowball Stemming
Preferred when you are working on multi-lingual text.




```python
print(filtered_words_without_special_characters)
print()

porter_stemmer = PorterStemmer()
prter_stemmed_words = [porter_stemmer.stem(word) for word in filtered_words_without_special_characters]
print(prter_stemmed_words)
print()

lancaster_stemmer = LancasterStemmer()
lancaster_stemmed_words = [lancaster_stemmer.stem(word) for word in filtered_words_without_special_characters]
print(lancaster_stemmed_words)
print()
```

## 1.4. Text Normalization: Lemmatization (do either stemming or lemmatization)
Lemmatization is the process of reducing the words to their dictionary form (lemma) while considering their context and part of speech.

- Running -> Run
- Studies -> Study
- Better -> Good

**Stemming** makes the words (improve, improving, improved, improvements) --> improv

**Lemmatization** makes the words (improve, improving, improved, improvements) --> improve

**Lemmatization** improves searching engines, sentiment analysis, Chatbot and Translation.

```python
print(filtered_words_without_special_characters)
print()

# Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words_without_special_characters]
print(lemmatized_words)
print()
```

# 2. EDA
In this example, EDA may not relevant. It is typically performed to do the followings:
- Find Text Lenght Distribution (no of words in a sentence)
- Word Frequency Analysis - to find out domain-specific terms that may need special handling.
- Class Balancing - Check distribution of words.
- Find out misspellings and Normalization needs.
- Sentiment (target) distribution

**EDA** does not change text into model features. Instead, it reveals dataset issues, and guides feature decissions.

# 3. Representation
In this stage, the input Text tokens are converted into numerical representations for models. Sometimes, you may have to add additional features like POS and NER to the input text before embedding.


## 3.1. Part of Speech (POS) Tagging
POS tagging is a fundamental technique in NLP that assigns grammatical labels - such as noun, verb, adjective, adverbs, pronouns conjuction, symbol, punctuation or preposition - to individual words within a sentense.

```python
# Resource for POS tagging for english language
nltk.download('averaged_perceptron_tagger_eng')

tokens_with_pos = nltk.pos_tag(lemmatized_words)
print(tokens_with_pos)
print()
```

## 3.2 Named Entity Recognition (NER)

```python
# Apply NER using POS tags and pre_trained models from nltk
from nltk import ne_chunk
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')

ner_tagged = ne_chunk(tokens_with_pos)
print(ner_tagged)
print()
```
