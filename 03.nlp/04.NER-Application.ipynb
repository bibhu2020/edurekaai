# 📖 Named Entity Recognition (NER)

## 🔹 What is NER?
- **Named Entity Recognition (NER)** is an **NLP task** where we identify and classify **real-world entities** in text into predefined categories.  
- Entities are usually **nouns or proper nouns** that represent important information.  
- Sometimes NER is a task in the **Representation** step in the NLP pipeline where you use is it to add additional feature information to the input.
- Sometimes NER is the final task or an application.

---

## 🔹 Example

Sentence:  
**"Apple is looking to buy a startup in New York for $1 billion."**

NER output:  
- `Apple` → **Organization**  
- `New York` → **Location**  
- `$1 billion` → **Monetary Value**  

---

## 🔹 Common Entity Types
- **Person** → "Barack Obama", "Elon Musk"  
- **Organization** → "Google", "United Nations"  
- **Location** → "Paris", "India"  
- **Date/Time** → "Monday", "2025-09-06"  
- **Money** → "$100", "5 euros"  
- **Percentages** → "50%"  
- **Products / Events / Medical terms** (depending on domain-specific NER)  

---

## 🔹 How NER Works
- Uses **linguistic rules + machine learning** to detect entities.  
- Modern systems rely on:
  - **Statistical models** (e.g., Conditional Random Fields).  
  - **Neural networks** (BiLSTM, Transformers like BERT).  
- Embeddings capture context, helping the model decide if "Apple" is a fruit 🍎 or a company 💻.  

---

## 🔹 Where is NER Used?

1. **Search Engines**  
   - Understand queries better by detecting entities.  
   - Example: `"Hotels in Paris"` → Entity: `"Paris" (Location)`.

2. **Chatbots & Virtual Assistants**  
   - Extract intent and key entities.  
   - `"Book a flight to London tomorrow"` → `Location=London`, `Date=tomorrow`.

3. **Information Extraction**  
   - Extracting **people, organizations, dates** from news articles or documents.

4. **Healthcare (Clinical NER)**  
   - Extract diseases, symptoms, drugs from medical notes.  
   - Example: `"Patient diagnosed with diabetes and prescribed insulin"` → `Disease=diabetes`, `Drug=insulin`.

5. **Finance**  
   - Identify companies, stock symbols, monetary values in financial reports.  

6. **Legal / Compliance**  
   - Automatically detect names, dates, case numbers in legal documents.  

---

## ✅ Quick Summary
- **NER = Find and label real-world entities in text**.  
- Example: `"Elon Musk founded SpaceX in 2002"` → `Person=Elon Musk`, `Organization=SpaceX`, `Date=2002`.  
- Used in **search, chatbots, healthcare, finance, legal, business intelligence**.  


# 📖 Statistical NER vs Neural NER

## 🔹 1. Statistical NER
- **Approach**: Based on **handcrafted features** + **probabilistic models**.  
- Common models:
  - **Hidden Markov Models (HMMs)**
  - **Conditional Random Fields (CRFs)**
  - **Maximum Entropy Markov Models (MEMMs)**
- Relies heavily on:
  - Part-of-Speech tags
  - Capitalization (e.g., "London" starts with a capital letter)
  - Word shapes (e.g., "Dr.", "Inc.")
  - Surrounding context words
- **Pros**:
  - Works well with small datasets.
  - More interpretable (you can see which features matter).
- **Cons**:
  - Needs **manual feature engineering**.
  - Struggles with unseen words or new domains.
  - Limited ability to capture **long-range context**.

---

## 🔹 2. Neural NER
- **Approach**: Uses **deep learning** models and embeddings.  
- Common models:
  - **BiLSTM + CRF** (classic neural NER architecture)
  - **Transformers (BERT, RoBERTa, BioBERT, ClinicalBERT)**
- Learns features automatically from:
  - Word embeddings (Word2Vec, GloVe, FastText)
  - Contextual embeddings (BERT, GPT)
- **Pros**:
  - Captures **contextual meaning** (e.g., "Apple" = fruit vs company).
  - No need for manual feature engineering.
  - Performs very well on large datasets.
  - Generalizes better to unseen words (especially with subword embeddings like BERT/FastText).
- **Cons**:
  - Requires **large annotated datasets**.
  - More **computationally expensive** (GPU often needed).
  - Harder to interpret.

---

## 🔹 Example
Sentence: `"Amazon is hiring in Paris."`  

- **Statistical NER** (feature-based):  
  - "Amazon" → Proper noun, capitalized → **Company (Organization)**  
  - "Paris" → Capitalized, appears in gazetteer list → **Location**  

- **Neural NER** (contextual embeddings):  
  - "Amazon" → Embedding context says it’s related to "hiring", "jobs" → **Organization**  
  - If sentence was `"Amazon river is flooding"`, neural NER would instead classify `"Amazon"` as a **Location (River)**.  

---

## ✅ Summary
| Aspect              | Statistical NER                   | Neural NER                        |
|---------------------|-----------------------------------|-----------------------------------|
| Features            | Handcrafted (POS, capitalization) | Learned automatically via embeddings |
| Models              | HMM, CRF, MEMM                   | BiLSTM-CRF, BERT, Transformers    |
| Data requirement    | Small                            | Large annotated corpora           |
| Context handling    | Limited (local window)            | Strong (long-range context)       |
| Accuracy            | Decent                           | State-of-the-art                  |
| Interpretability    | Easier                           | Harder                            |

# 🔎 spaCy vs StanfordNLP (now called Stanza)

Natural Language Processing (NLP) frameworks provide tools for tasks like tokenization, part-of-speech tagging, dependency parsing, and Named Entity Recognition (NER). Two of the most widely used libraries are **spaCy** and **StanfordNLP (Stanza)**. The latest libraries are built on neural network / deep learning.

---

## 📌 1. spaCy

- **Developer:** Explosion AI (Germany).  
- **Focus:** Industry-ready, fast, production-level NLP.  
- **Key Features:**
  - Tokenization, POS tagging, Dependency parsing, **NER**.
  - Pre-trained pipelines: `en_core_web_sm`, `en_core_web_md`, `en_core_web_lg`.
  - Integration with **Transformers** (BERT, RoBERTa, etc.).
  - Very fast and memory-efficient.
  - CNN based
- **Languages:** ~20+ supported.  
- **Best for:** Production systems, real-time NLP, pipeline customization.  

**Example: NER in spaCy**
```python
import spacy

nlp = spacy.load("en_core_web_sm")
doc = nlp("Apple is looking at buying a startup in New York for $1 billion.")

for ent in doc.ents:
    print(ent.text, ent.label_)
```

**Output:**
```
Apple ORG
New York GPE
$1 billion MONEY
```

---

## 📌 2. StanfordNLP (Stanza)

- **Developer:** Stanford NLP Group (USA).  
- **Focus:** Research-oriented, multilingual NLP with strong accuracy.  
- **Key Features:**
  - Tokenization, Lemmatization, POS tagging, Dependency parsing, **NER**.
  - Built on **PyTorch**.
  - Pre-trained models for **66+ languages**.
  - Models trained on **Universal Dependencies (UD)** + NER datasets.
  - LSTM-CRF based
- **Languages:** 66+ (much broader than spaCy).  
- **Best for:** Research, multilingual NLP tasks, academic projects.  

**Example: NER in Stanza**
```python
import stanza

stanza.download('en')   # first time only
nlp = stanza.Pipeline('en')

doc = nlp("Barack Obama was born in Hawaii and served as President of the United States.")

for ent in doc.entities:
    print(ent.text, ent.type)
```

**Output:**
```
Barack Obama PERSON
Hawaii GPE
President TITLE
United States GPE
```

---

## ⚖️ Comparison Table

| Feature              | spaCy 🟢                        | StanfordNLP / Stanza 🔵              |
|----------------------|---------------------------------|--------------------------------------|
| **Speed**            | Very fast, optimized for prod   | Slower, heavier models               |
| **Accuracy**         | High, especially in English     | State-of-the-art, strong accuracy    |
| **Languages**        | ~20+                           | 66+                                  |
| **NER Support**      | Yes, general-purpose            | Yes, general + multilingual          |
| **Ease of Use**      | Easy setup, lightweight         | Slightly heavier, needs PyTorch      |
| **Use Case**         | Production, industry            | Research, multilingual NLP           |

---

## ✅ Summary

- Use **spaCy** if you want **fast, production-ready NLP pipelines** (especially for English).  
- Use **Stanza (StanfordNLP)** if you need **multilingual support or research-grade accuracy**.

````python
import spacy
from spacy import displacy

nlp = spacy.load("en_core_web_sm")

raw_text="""The Indian Space Research Organisation or is the national
space agency of India, headquartered in Bengaluru.
It operates under Department of Space which is
directly overseen by the Prime Minister Narendra Modi of
India while Chairman of ISRO acts as executive of DOS as well."""

doc = nlp(raw_text)

for ent in doc.ents:
    print(ent.text, ent.label_)

displacy.render(doc,style='ent')
````

````python
import stanza

# Download English model (only needed the first time)
stanza.download('en')

# Initialize pipeline
nlp = stanza.Pipeline('en')

# Process text
doc = nlp("Barack Obama was born in Hawaii and served as President of the United States.")

# Print named entities
for ent in doc.ents:
    print(ent.text, ent.type)
````

### Transformer Based NER

The Transformer architecture is indeed the latest major neural network architecture that has revolutionized AI, especially in NLP and now also in vision, speech, and multimodal AI.

Transformers replaced older sequence models like RNNs, LSTMs, GRUs because of their efficiency and ability to capture long-range dependencies via self-attention.

BERT(Bidirectional Encoder Representations from Transformer) for NER

````python
from transformers import pipeline  # Hugging Face Transformers library

# Load pre-trained NER model pipeline
ner_pipeline = pipeline(
    task="token-classification",
    model="dslim/bert-base-NER",
    aggregation_strategy="simple"
)

# Sample texts
raw_texts = [
    "Barack Obama was born in Hawaii.",
    "Apple Inc. reported a revenue of $120 billion in 2024.",
    "Elon Musk founded SpaceX in California in 2002.",
    "Cristiano Ronaldo plays football in Saudi Arabia."
]

# Apply NER
for text in raw_texts:
    print(f"Text: {text}")
    print("Entities:", ner_pipeline(text))
    print("-" * 60)
````
