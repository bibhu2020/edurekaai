# Hugging Face Transformers `pipeline`

Yes ✅ — `from transformers import pipeline` is from **Hugging Face’s Transformers library**.

## What it does

- **`transformers`** is the Python library for working with state-of-the-art NLP models (BERT, GPT, T5, etc.).
- **`pipeline`** is a **high-level API** that makes it easy to perform common tasks without writing low-level model code.

## Example Tasks

| Task | Example |
|------|---------|
| Text Generation | `pipeline("text-generation", model="gpt2")` |
| Question Answering | `pipeline("question-answering", model="deepset/roberta-base-squad2")` |
| Sentiment Analysis | `pipeline("sentiment-analysis")` |
| Translation | `pipeline("translation_en_to_fr")` |
| Summarization | `pipeline("summarization")` |

## How it works

Under the hood, `pipeline`:

- Loads the pre-trained model and tokenizer.
- Handles preprocessing (e.g., tokenization) for you.
- Runs inference on the model.
- Decodes the model outputs back into human-readable text.

---

If you want, you can also use a GPT pipeline to answer questions **without providing a context**, generating text directly like ChatGPT.


## Extractive Transfer (has Encoder only)
It is good for Q&A, classification,....

Because it is extractive, we must pass the context to it.


```python
#Q & A BERT Pipeline
# qa_bert = pipeline("question-answering", model="deepset/bert-base-cased-squad2")
qa_bert = pipeline("question-answering", model="deepset/roberta-base-squad2")

#questions to ask
questions=[
    "what is machine learning?",
    "does supervised learning improve the accuracy of the model?",
    "which type of machine learning uses lablled data?",
    "where is the machine learning heading in the near future?",
    "what is deep learning?",
    "what is the future of machine learning?",
    "what is transformer?",
    "what is the difference between machine learning and deep learning?",
    "what is the difference between supervised and unsupervised learning"
]


for question in questions:
    answer = qa_bert(question=question, context=context)
    print(f"Question: {question}")
    print(f"Answer: {answer['answer']}")
```

## Generative Transformer (has Decoder)
It is used for text generation, translatio, summerization, etc...

```python
from transformers import pipeline

# Initialize a text-generation pipeline
generator = pipeline("text-generation", model="gpt2")  # or another GPT model

# Your prompt
prompt = "What is machine learning?"

# Generate text
output = generator(prompt, max_new_tokens=100)  # generates 100 new tokens after prompt

# Print generated text
print("Generated text:")
print(output[0]['generated_text'])

```

# Using OpenAI Models

## Text Classification

```python
from openai import OpenAI
client = OpenAI()

#Text Classification - Sentiment Analysis
text = "I love this ambience of the resturant,but the service is poor."
response = client.chat.completions.create(
    model="gpt-3.5-turbo",
    messages=[
        {"role": "system", "content": "You are a sentiment analysis assistant. Classify the sentiment as positive, neutral, or negative."},
        {"role": "user", "content": f"Classify the sentiment of the following text: '{text}'"}
    ]
    )
print(response.choices[0].message.content)
#
```

## Text Summerization

```python
from transformers import pipeline

# Initialize a summarization pipeline
summarizer = pipeline("summarization", model="facebook/bart-large-cnn")

# Input text to summarize
text = """
Machine learning is a field of artificial intelligence that focuses on building systems
that learn from data and improve their performance over time. Deep learning, a subfield
of machine learning, uses neural networks with many layers to automatically learn
features from data. These techniques are applied in various domains, including
computer vision, natural language processing, and robotics. As technology evolves,
AI continues to transform industries and create new opportunities.
"""

# Generate summary
summary = summarizer(text, max_length=60, min_length=25, do_sample=False)

# Print summarized text
print("Summary:")
print(summary[0]['summary_text'])
```

```python
from openai import OpenAI

client = OpenAI()

text = """
Machine learning is a field of artificial intelligence that focuses on building systems
that learn from data and improve their performance over time. Deep learning, a subfield
of machine learning, uses neural networks with many layers to automatically learn
features from data. These techniques are applied in various domains, including
computer vision, natural language processing, and robotics. As technology evolves,
AI continues to transform industries and create new opportunities.
"""

# Call OpenAI API to summarize
response = client.chat.completions.create(
    model="gpt-3.5-turbo",
    messages=[
        {"role": "system", "content": "You are a helpful assistant that summarizes text."},
        {"role": "user", "content": f"Please summarize the following text:\n\n{text}"}
    ],
    max_tokens=100,
    temperature=0.5
)

# Correct way to extract the generated text
summary = response.choices[0].message.content
print("Summary:")
print(summary)
```

## Translation of Content

```python
from openai import OpenAI

# Initialize the client
client = OpenAI()

# Text to translate
text_to_translate = """
Machine learning is a field of artificial intelligence that focuses on building systems
that learn from data and improve their performance over time.
"""

# Call OpenAI API to translate
response = client.chat.completions.create(
    model="gpt-3.5-turbo",  # or gpt-4 if available
    messages=[
        {"role": "system", "content": "You are a helpful assistant that translates English text to Hindi."},
        {"role": "user", "content": f"Translate the following text into Hindi:\n\n{text_to_translate}"}
    ],
    max_tokens=200,
    temperature=0.3
)

# Extract and print the translated text
translated_text = response.choices[0].message.content
print("Translated Text in Hindi:")
print(translated_text)
```
