## Prompt Templates
- Allows dynamic content with variable inputs.
- Consistent formatting.
- Ease of experiment
- Reusability
- Customization

**String Prompt**
A single-line with variables embedded in it.

E.g.
Tell me a joke about {topic}.

**Chat Prompt** (used for roles)
Organizes messages by roles (system, assistant, user).
Variables can be added to each message.
Ideal for simulating realistic coversations.

**Message Placeholder** (used for message history)
Adds dynamics message history into template.
Used in multi-turn, context-aware interactions.

```python
def make_prompt(topic, audience):
    template = (
        "You are an expert educator. "
        "Explain {topic} clearly for a {audience}."
    )
    return template.format(topic=topic, audience=audience)

prompt = make_prompt("Deep Learning", "college student")
print(prompt)
```

```python
from openai import OpenAI
import os
from google.colab import userdata
os.environ["OPENAI_API_KEY"]=userdata.get('OPENAI_API_KEY')
client = OpenAI()
```

```python
response = client.chat.completions.create(
    model="gpt-4-turbo",  # or "gpt-3.5-turbo"
    messages=[
        {"role": "user", "content": prompt}
    ]
)

# Print model output
print(response.choices[0].message.content)
```

```python
prompt = f"""
You are a legal expert. Explain the following clause in simple English:
"{'The party of the first part shall indemnify the second part...'}"
"""

response = client.chat.completions.create(
    model="gpt-4",
    messages=[
        {"role": "system", "content": "You are an expert legal assistant."},
        {"role": "user", "content": prompt}
    ]
)

print(response.choices[0].message.content)
```

```python
def domain_qa(domain, context, question):
    TEMPLATE = f"""
    You are a domain expert in {domain}.
    Use the provided context to answer the question clearly, factually, and concisely.

    Context:
    {{context}}

    Question:
    {{question}}

    Answer:
    """
    prompt = TEMPLATE.format(context=context, question=question)

    response = client.chat.completions.create(
        model="gpt-4-turbo",
        messages=[{"role": "user", "content": prompt}],
        temperature=0.3,
    )
    return response.choices[0].message.content.strip()
```

```python
context = """
Hypertension, also known as high blood pressure, is a chronic condition in which
the blood pressure in the arteries is persistently elevated. It increases the risk
of heart disease, stroke, and kidney failure. Lifestyle modifications such as
reducing salt intake, regular exercise, and maintaining a healthy weight are
recommended for management.
"""

question = "What are the lifestyle changes recommended for controlling hypertension?"
```

```python
answer = domain_qa(
    "Medical",
    context,
    question
)

print(answer)
```

```python
response = client.chat.completions.create(
    model="gpt-4-turbo",
    messages=[
        {"role": "system", "content": "You are a medical expert."},
        {"role": "user", "content": answer}
    ],
    temperature=0.3,  # low temp for factual answers
)

print(response.choices[0].message.content)
```

**Using LangChain Prompt Template Class**


You define placeholders ({variable}), and LangChain fills them with runtime values before sending to the model.

```python
from langchain.prompts import PromptTemplate

# Create a basic prompt template
template = """
You are a helpful assistant that explains {topic} in simple terms.
Explain it to a {audience}.
"""

prompt = PromptTemplate(
    input_variables=["topic", "audience"],
    template=template,
)

# Generate the final prompt
filled_prompt = prompt.format(topic="Quantum Computing", audience="high school student")
print(filled_prompt)
```
