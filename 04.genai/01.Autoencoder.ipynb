## Autoencoder

**Autoencoders** represent one of the earliest forms of content-generating neural networks. They laid the foundation for more advanced architectures such as **Generative Adversarial Networks (GANs)**, **Small Language Models (SLMs)**, and **Large Language Models (LLMs)**.

---

### What is an Autoencoder?

An **autoencoder** is a type of neural network that learns to **compress data** (encode) into a smaller, efficient representation and then **reconstruct it** (decode) back to something close to the original input.  
This training occurs in an **unsupervised manner**, meaning the model learns patterns directly from the input data without labeled outputs.

---

### Structure of an Autoencoder

```
Input → Encoder → Latent Space (Encoded Features) → Decoder → Output
```

![autoencoder](https://www.compthree.com/images/blog/ae/ae.png)

Each component plays a unique role:

#### 1. **Encoder**
- The encoder’s job is to **compress** the input data into a smaller set of numerical features known as **latent features**.
- It typically consists of **convolutional** or **dense (fully connected)** layers that gradually reduce dimensionality.
- As the data flows through these layers, the encoder filters out redundant or irrelevant information.
- Example: When compressing an image, the encoder learns which pixels contribute most to identifying key objects, shapes, and patterns, while ignoring background noise.

#### 2. **Latent Space (Bottleneck Layer)**
- The **latent space** represents the **compressed knowledge** of the input — the “essence” of the data.
- It’s a lower-dimensional feature space that captures only the most important information.
- In images, it might store key textures, shapes, or color gradients.  
- In text, it may encode semantic meaning or relationships between words.

#### 3. **Decoder**
- The decoder takes the encoded latent features and **reconstructs** the data back into its original format.
- It performs an **inverse mapping** of the encoder’s transformation.
- While the goal is to make the output resemble the input, some loss of detail is inevitable due to compression.

---

### Training Process of an Autoencoder

1. **Input:** Feed the model raw data (e.g., images, text, or audio).  
2. **Encoding:** The encoder compresses the input into latent features. It is also called dimensionality reduction. (**PCA** (Principal Component Analysis) is used for linear data. **Encoder** is used for non-linear data in neural network)
3. **Decoding:** The decoder reconstructs the original data from those features. (lossy output)
4. **Loss Calculation:** The model measures how close the reconstructed output is to the original input (e.g., Mean Squared Error).  
5. **Weight Update:** Using backpropagation, it adjusts weights to minimize reconstruction loss.  
6. **Iteration:** Repeat over multiple epochs until the model converges.
[![Watch the video](https://img.youtube.com/vi/VIDEO_ID/maxresdefault.jpg)](https://www.youtube.com/watch?v=hZ4a4NgM3u0)

**_4 hype-parameters used for Training Autoencoder_**:
  - code_size: intended size of the latent feature dimension.
  - number_of_layers: autoencoder can have many layers.
  - loss_function: "binary-cross-entropy" is used for input value 0-1 or "mse" is used.
  - node_per_layers: No of neurons on each layer.

**Note**
 - In encoder, the layer size decreases to code_size.
 - In decoder, the layer size increases from code_size in the same size pattern.

---

### Real-World Applications of the Encoder

The **encoder** is the most crucial part of the autoencoder — and its learned features are widely used in practical applications:

#### 1. **Image Denoising**
- Encoders learn to remove unwanted noise from images.
- Example: Enhancing old or blurred photos by training the model to reconstruct a clean version.

#### 2. **Dimensionality Reduction**
- Autoencoders are used to reduce the number of input features while retaining important information.
- Example: Compressing high-dimensional sensor data before feeding it to another ML model.

#### 3. **Anomaly Detection**
- Encoders learn to reconstruct “normal” patterns in data.  
- When they fail to accurately reconstruct a new input, it signals an anomaly.
- Example: Detecting fraudulent transactions or equipment failures.

#### 4. **Recommendation Systems**
- The encoder learns compact representations of user behavior or content.
- Example: Movie or product recommendation systems (e.g., Netflix, Amazon) use these embeddings to find similar users or items.

#### 5. **Natural Language Processing (NLP)**
- Encoders form the basis of transformer models like **BERT**, where they learn context-rich embeddings of text.
- Example: Sentence classification, semantic search, and text similarity tasks.

#### 6. **Medical Imaging**
- Encoders are used to extract key patterns from MRI or CT scans for downstream tasks such as segmentation or disease classification.

---

### Pros and Cons of Autoencoders

**Pros**
- Excellent for **denoising** and **feature extraction**.  
- Helps with **dimensionality reduction** for visualization or preprocessing.  
- Learns meaningful **latent representations** without labels.  

**Cons**
- **Loss of detail** during compression can lead to imperfect reconstructions.  
- Not ideal for generating **high-quality new data**, since it focuses on reproducing existing inputs.  

---

**Autoencoder** is intended for producing the same output. **Seq2Seq** model is used for producing different output (e.g. content translation).

They both are simillar in architecture, the difference is in its training. In Autoencoder, the output is compared with the input.

Loss=L(input,output)

In Seq2Seq, the output is compared with a different output.

Loss=L(target sequence,predicted sequence)

**Autoencoder** is used for Denoising (removing blur from image, clearing up speech), Anomaly Detection (when input does not fit a pattern, the reconstruction error spikes), and Data Compression.

**Seq2Seq** models are used for Translation, Text Summerization, QA, Chatbot, Speech Recognition, Text-to-Speech, Code Generation, Rewriting, and Timeseries forecasting.




## From Autoencoders to Variational Autoencoders (VAEs)

The **latent space** in a standard autoencoder is smaller in dimension than the input, which means some information is inevitably lost during compression.  
This loss restricts the ability to generate new, diverse outputs.

The Problems:
- No understanding of data distribution
- Latent space is not smooth or continious

To address this, researchers developed **Variational Autoencoders (VAEs)** — a probabilistic extension that learns to sample from a continuous latent distribution.  
VAEs enable **controlled data generation**, producing more realistic and varied outputs while maintaining structure and consistency.

## Understanding Variational Autoencoders (VAEs)

Autoencoders are a powerful type of neural network that can compress data into a smaller representation and then reconstruct it. But regular autoencoders have a limitation: their latent representations are fixed points, which makes generating new data or exploring “what-if” variations tricky. **Variational Autoencoders (VAEs)** solve this problem by introducing probability into the latent space. Let’s explore what this means and why it matters.

---

### 1️⃣ What is a latent vector?

When an autoencoder processes an input — say, an image of a cat — the encoder compresses it into a **latent vector**, a smaller representation that captures the essential features of the input.

Example:

```
Input: Image of a cat
Latent vector: [0.2, 0.9, -0.1]
```

This latent vector is then used by the decoder to reconstruct the original input.

---

### 2️⃣ The limitation of fixed latent points

In a **regular autoencoder**, each input is mapped to **exactly one point** in latent space. This works for reconstruction, but it has a drawback:

* If you **slightly tweak the latent vector**, the decoder may fail to produce a meaningful output.

**Why does this happen?**

* The decoder is trained only on the exact latent vectors produced by the encoder.
* It has **never seen nearby points** during training.
* So, moving even a little from the known point can result in gibberish output.

#### Analogy

Imagine the latent space as a map with cities:

* The decoder knows the roads only between **the cities it has seen**.
* If you try to place yourself **between cities**, there are no roads — you’re lost, and the decoder outputs nonsense.

---

###3️⃣ Enter Variational Autoencoders (VAEs)

VAEs solve this problem by representing the latent space **probabilistically** rather than as a fixed point.

* Instead of encoding an input as a single vector, the encoder outputs **two vectors**:

  1. **Mean (μ):** the center of the latent distribution
  2. **Variance (σ²):** how spread out the distribution is

* This defines a **Gaussian distribution** in latent space for each input.

![VAE](https://www.mdpi.com/make/make-02-00020/article_deploy/html/images/make-02-00020-g001-550.jpg)

---

#### 4️⃣ How sampling works in VAEs

During training, the model doesn’t just use the mean. It **samples from the Gaussian distribution** defined by (μ, σ²):

```
Sampled z = μ + σ * ε,   where ε ~ N(0,1)
```

* This forces the decoder to handle a **range of nearby points**, not just a single fixed vector.
* As a result, the latent space becomes **continuous and smooth**.

---

### 5️⃣ Why this matters

* **Robust reconstruction:** Small changes in the latent vector still produce meaningful outputs.
* **Generative ability:** You can sample from the latent space to generate new, realistic data.
* **Interpolation:** You can morph one image into another smoothly by moving through latent space.

---

#### 6️⃣ Visual analogy

* **Regular autoencoder:** Each input maps to a single dot in latent space. Nearby points may produce nonsense.
* **VAE:** Each input maps to a cloud of points (distribution). The decoder is trained on these points, so outputs are meaningful even for small variations.

```
Latent Space (1D example)
Regular AE:       o       o       o   (dots for each input)
VAE:             ooo     ooo     ooo   (clouds for each input)
```

---

### 7️⃣ Key takeaway

* **Regular autoencoders**: deterministic, can reconstruct known inputs, not reliable for generating new data.
* **VAEs**: probabilistic, learn distributions in latent space, capable of generating new, realistic outputs.

The main innovation of VAEs is that they allow **flexible, continuous, and generative latent spaces**, turning the autoencoder from a simple compressor into a powerful generative model.

### Reconstruct a 10×10 Grayscale Image using Autoencoder

````python
# ============================================
# 🧠 Simple Autoencoder (AE)
#  Step-by-Step Vector Flow Visualization
#  Using a 10×10 Grayscale Image
# ============================================

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Model
import matplotlib.pyplot as plt

# ------------------------------------------------
# 1️⃣ Create a simple 10×10 grayscale image
# ------------------------------------------------
image = np.zeros((10, 10))
image[2:8, 3:7] = 1.0  # bright block in the center

plt.imshow(image, cmap='gray')
plt.title("Original 10×10 Image")
plt.axis('off')
plt.show()

# Flatten input
x_train = image.flatten()[np.newaxis, :]
print("\n🔹 Input Flattened Vector (x_train):")
print(x_train)

# ------------------------------------------------
# 2️⃣ Build Autoencoder
# ------------------------------------------------
input_dim = 100
latent_dim = 1  # 1D latent for comparison with VAE

# Encoder Model
encoder_inputs = layers.Input(shape=(input_dim,), name="encoder_input")
encoder_hidden = layers.Dense(32, activation="relu", name="encoder_hidden")(encoder_inputs)
latent = layers.Dense(latent_dim, activation=None, name="latent")(encoder_hidden)
encoder = Model(encoder_inputs, [encoder_hidden, latent], name="encoder")

# Decoder Model
latent_inputs = layers.Input(shape=(latent_dim,), name="latent_input")
decoder_hidden = layers.Dense(32, activation="relu", name="decoder_hidden")(latent_inputs)
decoder_outputs = layers.Dense(input_dim, activation="sigmoid", name="decoder_output")(decoder_hidden)
decoder = Model(latent_inputs, [decoder_hidden, decoder_outputs], name="decoder")

# Autoencoder
ae_inputs = encoder_inputs
_, latent_vec = encoder(ae_inputs)
_, ae_outputs = decoder(latent_vec)
autoencoder = Model(ae_inputs, ae_outputs, name="autoencoder")

# ------------------------------------------------
# 3️⃣ Compile and Train
# ------------------------------------------------
autoencoder.compile(optimizer="adam", loss="mse")
autoencoder.fit(x_train, x_train, epochs=300, batch_size=1, verbose=0)
print("✅ Training complete")

# ------------------------------------------------
# 4️⃣ Extract Intermediate Outputs
# ------------------------------------------------
encoder_hidden_val, latent_val = encoder.predict(x_train)
decoder_hidden_val, reconstruction_val = decoder.predict(latent_val)

# ------------------------------------------------
# 5️⃣ Print Vector Transformations
# ------------------------------------------------
np.set_printoptions(precision=4, suppress=True)

print("\n================== 🧩 ENCODER STAGE ==================")
print("Input Vector (x_train):")
print(x_train)

print("\nEncoder Hidden Layer Output (32-D):")
print(encoder_hidden_val)

print("\nLatent Vector (1-D):")
print(latent_val)

print("\n================== 🔁 DECODER STAGE ==================")
print("Decoder Hidden Layer Output (32-D):")
print(decoder_hidden_val)

print("\nReconstructed Output Vector (100-D):")
print(reconstruction_val)

# ------------------------------------------------
# 6️⃣ Visualize Reconstructed Image
# ------------------------------------------------
reconstructed_image = reconstruction_val.reshape(10, 10)
plt.imshow(reconstructed_image, cmap="gray")
plt.title("🖼️ Reconstructed 10×10 Image (Autoencoder)")
plt.axis("off")
plt.show()

print("\n✅ AE visualization complete. Compare with VAE outputs to see differences.")
````

### Reconstruct a 10×10 Grayscale Image using VAE

````python
# ============================================
# 🧠 Variational Autoencoder (VAE)
#  Step-by-Step Vector Flow Visualization
#  Using a 10×10 Grayscale Image
# ============================================

import numpy as np
import tensorflow as tf
from tensorflow.keras import layers, Model
import matplotlib.pyplot as plt

# ------------------------------------------------
# 1️⃣ Create a simple 10×10 grayscale image
# ------------------------------------------------
image = np.zeros((10, 10))
image[2:8, 3:7] = 1.0  # bright block in the center

plt.imshow(image, cmap='gray')
plt.title("Original 10×10 Image")
plt.axis('off')
plt.show()

# Flatten and normalize
x_train = image.flatten()[np.newaxis, :]  # shape (1, 100)
print("\n🔹 Input Flattened Vector (x_train):")
print(x_train)

# ------------------------------------------------
# 2️⃣ Define the Sampling Layer (μ + σ * ε)
# ------------------------------------------------
class Sampling(layers.Layer):
    """Reparameterization trick: z = μ + σ * ε"""
    def call(self, inputs):
        mu, log_sigma = inputs
        epsilon = tf.random.normal(shape=tf.shape(mu))
        return mu + tf.exp(0.5 * log_sigma) * epsilon

# ------------------------------------------------
# 3️⃣ Build Encoder and Decoder
# ------------------------------------------------
input_dim = 100
latent_dim = 1

# Encoder definition
encoder_inputs = layers.Input(shape=(input_dim,), name="encoder_input")
h = layers.Dense(32, activation="relu", name="encoder_hidden")(encoder_inputs)
mu = layers.Dense(latent_dim, name="mu")(h)
log_sigma = layers.Dense(latent_dim, name="log_sigma")(h)
z = Sampling(name="z")([mu, log_sigma])

encoder = Model(encoder_inputs, [h, mu, log_sigma, z], name="encoder")

# Decoder definition
latent_inputs = layers.Input(shape=(latent_dim,), name="z_sampling")
decoder_hidden = layers.Dense(32, activation="relu", name="decoder_hidden")(latent_inputs)
decoder_outputs = layers.Dense(input_dim, activation="sigmoid", name="decoder_output")(decoder_hidden)

decoder = Model(latent_inputs, [decoder_hidden, decoder_outputs], name="decoder")

# ------------------------------------------------
# 4️⃣ Combine into VAE Model
# ------------------------------------------------
class VAE(Model):
    def __init__(self, encoder, decoder, **kwargs):
        super(VAE, self).__init__(**kwargs)
        self.encoder = encoder
        self.decoder = decoder

    def train_step(self, data):
        if isinstance(data, tuple):
            data = data[0]
        with tf.GradientTape() as tape:
            h, mu, log_sigma, z = self.encoder(data)
            dec_h, reconstruction = self.decoder(z)

            # Compute losses
            reconstruction_loss = tf.reduce_mean(tf.reduce_sum(tf.square(data - reconstruction), axis=1))
            kl_loss = -0.5 * tf.reduce_mean(
                tf.reduce_sum(1 + log_sigma - tf.square(mu) - tf.exp(log_sigma), axis=1)
            )
            total_loss = reconstruction_loss + kl_loss

        grads = tape.gradient(total_loss, self.trainable_weights)
        self.optimizer.apply_gradients(zip(grads, self.trainable_weights))
        return {"loss": total_loss, "reconstruction_loss": reconstruction_loss, "kl_loss": kl_loss}

# ------------------------------------------------
# 5️⃣ Compile and Train the Model
# ------------------------------------------------
vae = VAE(encoder, decoder)
vae.compile(optimizer=tf.keras.optimizers.Adam())
vae.fit(x_train, x_train, epochs=300, batch_size=1, verbose=0)
print("\n✅ Training complete")

# ------------------------------------------------
# 6️⃣ Extract Intermediate Outputs for Visualization
# ------------------------------------------------
# Pass input through encoder
h_val, mu_val, log_sigma_val, z_val = encoder.predict(x_train)
sigma_val = np.exp(0.5 * log_sigma_val)

# Pass latent through decoder
decoder_hidden_val, reconstruction_val = decoder.predict(z_val)

# ------------------------------------------------
# 7️⃣ Print Vector Transformations
# ------------------------------------------------
np.set_printoptions(precision=4, suppress=True)

print("\n================== 🧩 ENCODER STAGE ==================")
print("Input Vector (x_train):")
print(x_train)

print("\nEncoder Hidden Layer Output (32-D):")
print(h_val)

print("\nLatent Mean (μ):")
print(mu_val)

print("\nLatent Log Variance (log σ²):")
print(log_sigma_val)

print("\nLatent Std Dev (σ):")
print(sigma_val)

print("\nSampled Latent Vector (z = μ + σ·ε):")
print(z_val)

print("\n================== 🔁 DECODER STAGE ==================")
print("Decoder Hidden Layer Output (32-D):")
print(decoder_hidden_val)

print("\nReconstructed Output Vector (100-D):")
print(reconstruction_val)

# ------------------------------------------------
# 8️⃣ Visualize Reconstructed Image
# ------------------------------------------------
reconstructed_image = reconstruction_val.reshape(10, 10)

plt.imshow(reconstructed_image, cmap="gray")
plt.title("🖼️ Reconstructed 10×10 Image")
plt.axis("off")
plt.show()

print("\n✅ Visualization complete. You can now inspect how data changes layer by layer.")

# ------------------------------------------------
# 9️⃣ Generate Multiple Samples from the Latent Space
# ------------------------------------------------
num_samples = 10  # how many random samples to generate

# Use the learned latent mean and std from the encoder
mu_val = mu_val[0]
sigma_val = sigma_val[0]

generated_images = []

for i in range(num_samples):
    # Sample z ~ N(μ, σ²)
    z_random = mu_val + sigma_val * np.random.randn(latent_dim)
    _, recon = decoder.predict(z_random[np.newaxis, :])
    generated_images.append(recon.reshape(10, 10))

# ------------------------------------------------
# 🔍 Visualize Multiple Generated Samples
# ------------------------------------------------
fig, axes = plt.subplots(1, num_samples, figsize=(15, 2))
for i in range(num_samples):
    axes[i].imshow(generated_images[i], cmap='gray')
    axes[i].set_title(f"Sample {i+1}")
    axes[i].axis("off")

plt.suptitle("🎨 Multiple Samples from VAE Latent Space")
plt.show()

````
