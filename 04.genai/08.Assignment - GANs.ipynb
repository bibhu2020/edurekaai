# GAN Assignment: Generate Handwritten Digits

## Objective
Learn the fundamentals of GANs by building a simple generator-discriminator network to generate handwritten digits similar to the MNIST dataset.

---

## Steps

### 1️⃣ Set Up Your Environment
- Use **Python** and **PyTorch** (or TensorFlow/Keras if you prefer).
- Install required packages:

```bash
pip install torch torchvision matplotlib tensorflow

```

---

### 2️⃣ Load Dataset
- Use MNIST dataset from `torchvision.datasets`.
- Normalize pixel values to [-1, 1] for better GAN performance.

---

### 3️⃣ Define Generator and Discriminator
- **Generator**: Takes random noise (latent vector) and outputs a 28x28 image.
- **Discriminator**: Takes an image (28x28) and outputs a probability that it’s real.

**Hints:**
- Generator: Use a few linear layers with `ReLU` and `tanh` at output.
- Discriminator: Use a few linear layers with `LeakyReLU` and `sigmoid` at output.

---

### 4️⃣ Define Loss and Optimizer
- Use **Binary Cross-Entropy Loss**.
- Use **Adam optimizer** for both generator and discriminator.

---

### 5️⃣ Train the GAN
- **For each epoch:**
  1. Train discriminator on real images and generated fake images.
  2. Train generator to fool the discriminator.
- Train for **at least 50–100 epochs**.

---

### 6️⃣ Generate Samples
- Use the trained generator to create new images from random noise.
- Visualize them using `matplotlib.pyplot`.

---

### 7️⃣ Test and Evaluate
- Check if the generated digits look realistic.
- Optional: Compare images from early epochs vs later epochs to see improvement.

---

## Deliverables
1. **Python notebook or script** including:
   - Data loading
   - Generator & discriminator definition
   - Training loop
   - Visualization of generated images
2. **Brief report** explaining:
   - Architecture of the generator and discriminator
   - Loss curves
   - Example outputs at different epochs
