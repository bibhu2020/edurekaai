## Why Small Language Model?
- SML has significantly lesser parameters (typically < 10B)
- Efficient and fast, optimized for specific use cases.
- Exampled: Phi-2, TinyLlama, Mistral-7B
- Requires lower compute and memory requirements.
- Faster inference times.
- Security concerns with LLM running on 3rd party server.
- LLM parameters are from billions to trillions.
- SLM paramertes from millions to 10 billions.
