## Long-Short Term Memory (LSTM)
LSTM is a special kind of RNN designed to lean long-term dependencies.
It solves the vanishing & exploding gradient problem that makes basic RNNs struggle with the long sequences.

Applications
- **Sentiment Analysis**
- Speech recognition
- time series
- text generation
- chatbot

LSTM uses 2 separate path to make prediction for tomorrow:
- One path for long-term memory (GREEN line below shows long-term memory)
- Another path for short-term memory (RED line represents short-term memory)

Let's understand this with the assumption that at the starting point, we have 2 in long-term memory and 1 in short-term memory from previous day.

Now, the new input from today that it is about to process is 1.

**Long-Term Memory Calculation**

**_Stage1 (Forget Gate)_**
- Stage1 is represented by the blue-box.
- It calculates how much % of the previous day memory is forgotten OR how much % is propagated to today.
- The blue-box uses sigmoid function to calculate the % (i.e. 0.997)
- The long-term memory that is propagated is 1.99 (2 x 0.997)

**_Stage2 (Input Gate)_**
- Stage2 is represented by the green-box and orange-box.
- orange-box creates the potential long-term memory of today's input. (uses sigmoid)
- green-box calculates what % of the created potential long-term memory shall be added to the long-term memory path to create the **new long-term memory**. (uses tanh)

**Short-Term Memory Calculation**

**_Stage3 (Output Gate)_**
- It is represented by the purple-box and red-box. It calculates the **new short-term memory**.
- The red-box uses tanh to calculate the potential short-term memory.
- The purple-box uses sigmoid to calculate the % of potential short-term memory to remember. It generates the **new short-term memory** or the **input to the output-layer**.
