# LLM Training
LLM was originally trained to perform the task of predicting the next word in an incomplete sentence. But It suprised its inventor by performing multiple other tasks that it was not trained for like Q&A, Classification, Translation, Content Generation etc..

## Pretraining vs Finetunning
**Pretraining** means training a LLM on a huge, and diverse dataset.It takes a huge computational power. (e.g. pre-training of GPT3 was ~4.6 million dollars). Pretraining output is called Foundational Model.

**Finetunning** is refinement "Foundational Model" by training on narrow dataset, specific to a particular task or domain.

## Quantization
**Analogy**: It is like storing numbers as 2 decimal points instead of 8 — good enough for most calculations, but much smaller and faster to process.

Normally, model weights are stored in 32-bit floating point (FP32) format. **Quantization** converts these to lower-precision formats, e.g.,:
- 16-bit floating point (FP16)
- 8-bit integer (INT8)
- 4-bit integer (INT4)

# Fine Tunning Techniques

This document explains various fine-tuning techniques for modern LLMs using BERT Base (110M params) and BERT Large (~340M params) as examples.

| Model          | Encoder Layers | Hidden Size (`d_model`) | Attention Heads | FFN Size (`d_ff`) | Parameters |
| -------------- | -------------- | ----------------------- | --------------- | ----------------- | ---------- |
| **BERT Base**  | 12             | 768                     | 12              | 3072              | ~110M      |
| **BERT Large** | 24             | 1024                    | 16              | 4096              | ~340M      |


## 1. Full Parameter Fine-Tuning (FPFT)

* **How it works:** Update **all parameters** of the model.
* **BERT Base:** ~110M parameters → all trained.
* **BERT Large:** ~340M parameters → all trained.
* **Implications:**
  * Requires lots of GPU memory and compute.
  * Risk of overfitting on small datasets.
* **Use case:** Large dataset, need full adaptation.

## 2. Quantization-Aware Fine-Tuning

* **Concept:** Train/adapt the model in **lower precision** (8-bit, 4-bit).
* **Example:**

  * BERT Large (~340M) quantized to 8-bit → reduces memory from ~1.3 GB → ~0.3 GB during inference.
* **Benefit:** Saves compute and memory while still adapting the model.

## 3. Parameter-Efficient Fine-Tuning (PEFT)

### a. Adapter Tuning

* **Concept:** Add tiny adapter modules between layers.
* **Example:**

  * BERT Base: ~110M → adapter adds ~2–3M trainable params.
  * BERT Large: ~340M → adapter adds ~7–10M trainable params.
* **Pros:** Swap adapters for different tasks.

### b. LoRA (Low-Rank Adaptation)

* **Concept:** Injects small trainable matrices Adapter into **attention layers**.
* **Effect:** Instead of updating all parameters, only a small fraction is trained.
* **Pros:** Much cheaper than full fine-tuning; preserves pre-trained knowledge.

### c. QLoRA (Quantized LoRA)

* **Concept:** Combines **LoRA** with **4-bit quantization** of the base model.
* **Effect:** Train LoRA adapters on a **quantized frozen model**, dramatically reducing memory usage.
* **Pros:** Enables fine-tuning very large models on consumer GPUs; extremely efficient.
* **Use case:** Large-scale domain adaptation without heavy compute resources.

## Comparison of Fine-Tuning Techniques for BERT Large

| Technique                      | Trainable Params  | GPU/Memory Usage               | Efficiency                      | Pros                                                                           | Cons                                                                         |
| ------------------------------ | ----------------- | ------------------------------ | ------------------------------- | ------------------------------------------------------------------------------ | ---------------------------------------------------------------------------- |
| Full Fine-Tuning               | ~340M             | Very high (~16–32GB+ for FP32) | Low efficiency for large models | Maximum flexibility; full adaptation                                           | Expensive; risk of overfitting; high memory/computation cost                 |
| Quantization-Aware Fine-Tuning | ~340M (quantized) | Reduced (~0.3–0.5GB for 8-bit) | Medium                          | Saves memory and compute; allows adaptation on smaller GPUs                    | Slight precision loss; still trains full model                               |
| Adapter Tuning                 | ~7–10M            | Low (~1–2GB)                   | High                            | Task-specific; easy to swap adapters; lightweight                              | Limited model adaptation; may underperform full fine-tuning on complex tasks |
| LoRA                           | ~3–5M             | Low (~1–2GB)                   | Very high                       | Efficient; preserves pre-trained knowledge; trains small fraction              | May need careful rank selection; limited adaptation scope                    |
| QLoRA                          | ~3–5M             | Very low (~0.5–1GB)            | Extremely high                  | Combines LoRA + quantization; enables large-scale fine-tuning on consumer GPUs | Requires quantization-aware setup; may slightly affect accuracy              |
